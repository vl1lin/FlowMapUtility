# Идея для следующего проекта
Использовать уже имеющуюся утилиту, но вместо построения карты, определять режим течения в точке (для этого придется знать дебиты)

## Как это будет выглядеть
- Пользователь задает в координатах X, Y линию тракетории скважины


- алгоритм разбивает на наиболее интересные участки

    Сам алгоритм будет основываться на регресси, сначала поределяем крупные участки с определенным паттерном наклона, потом приближаем

  Вместо определения какая корреляция должна использоваться можно использовать корреляцию Barnea

- рассчитывает для них ражим течения, выводит массив: $\text{[X, Y, режим течения]}$


- дальше можно отдать визуализатору (построить график где будет линия траектории скважины, причем цвет линии будет показывать режим потока на этом участке)

- Также можно для участков строить карты режим течения предполагая границы скоростей вблизи рассчитанных из дебитов.
